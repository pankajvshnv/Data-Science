```python
data = pd.read_csv("sentiment_emotion_dataset_100.csv")
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
emotion_type = data['emotion_type'].unique()

emotion_dict = {}
no = 0
for i in emotion_type:
    emotion_dict[i] = no
    no = no +1
data['emotion_type'] = data['emotion_type'].map(emotion_dict)
```

```python
data.head()
```

```python
data['sentence'] = data['sentence'].str.lower()
```

```python
data
```

```python
import string

data['sentence'] =data['sentence'].str.translate(
    str.maketrans('', '', string.punctuation)
)
```

## Bag of Words
