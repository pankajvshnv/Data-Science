```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("data.csv")
```

```python
df = data
```

```python
df.info()
```

df.isnull().sum()

```python
df['Engine'] = df['Engine'].astype(str).str.replace(' cc', '').astype(float)
df = df.rename(columns={"Make":"Brand"})
```

```python
df.head()
```

```python

df['Model'] = df['Model'].astype(str).str.split().str[0]


df['Max Power'] = df['Max Power'].str.split('@').str[0].str.split().str[0].astype(float)
df['Max Torque'] = df['Max Torque'].str.split('@').str[0].str.split().str[0].astype(float)
```

```python
df = df.rename(columns={"Max Power":"Max Power bhp"})
```
